## Data Cleaning

```python
## Import libraries
import numpy as np
import pandas as pd
```

```python
## Load dataset
df = pd.read_csv("../data/raw/Telco-Customer-Churn.csv")

## First five observations
df.head()
```

```python
## Check total no of missing values in each column
df.isnull().sum()
```

### Change Data Types

```python
## Convert 'TotalCharges' column to numeric, forcing errors to NaN
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

## Check data type
df['TotalCharges'].dtype
```

### Handling Duplicate Rows

```python
## Check for duplicate rows in the dataset
df.duplicated().sum()
```

```python
## Drop duplicate rows
df.drop_duplicates(inplace=True)
```

### Handling Missing Values

```python
## Check total no of missing values in each column
df.isnull().sum()
```

```python
(df.isnull().sum()/len(df))*100
```

#### Comment:
- We can see that 3 columns contains missing values. 
- Dependents column contains 0.04% missing values, MonthlyCharges contains 0.07% and TotalCharges contains 0.19% missing values. 
- Since the percentage of missing values is very low, we can drop these rows without losing much information.

```python
missing_col = ['Dependents', 'MonthlyCharges', 'TotalCharges']

df.dropna(subset=missing_col, inplace=True)

df.isnull().sum()
```

#### Comment:
Now the dataset does not contain any missing values. So we can proceed with the analysis.

### Drop Unnecessary Columns

```python
## Drop customerId column
df.drop(columns=['customerID'], inplace=True)

df.head()
```

### Export Cleaned Data

```python
df.to_csv('../data/processed/Cleaned-Telco-Customer-Churn.csv', index=False)
```
